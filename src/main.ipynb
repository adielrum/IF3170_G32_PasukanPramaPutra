##### Categorical Correlation

```python
from scipy.stats import chi2_contingency

categorical_cols = [
    "meal_package",
    "room_category",
    "booking_channel",
    "needs_parking",
    "is_returning_guest"
]

def cramers_v(x, y):
    table = pd.crosstab(x, y)
    chi2 = chi2_contingency(table, correction=False)[0]
    n = table.to_numpy().sum()
    return np.sqrt((chi2 / n) / min(table.shape[0] - 1, table.shape[1] - 1))

def association_strength(v, k):
    thresholds = {
        1: (0.10, 0.30, 0.50),
        2: (0.07, 0.21, 0.35),
        3: (0.06, 0.17, 0.29),
        4: (0.05, 0.15, 0.25),
        5: (0.05, 0.13, 0.22)
    }

    weak, moderate, strong = thresholds[k]

    if v < weak:
        return "Negligible"
    elif v < moderate:
        return "Weak"
    elif v < strong:
        return "Moderate"
    else:
        return "Strong"
```

```python
results = []

for i in range(len(categorical_cols)):
    for j in range(i + 1, len(categorical_cols)):
        col1 = categorical_cols[i]
        col2 = categorical_cols[j]

        table = pd.crosstab(train_df[col1], train_df[col2])
        k = min(table.shape[0] - 1, table.shape[1] - 1)
        value = cramers_v(train_df[col1], train_df[col2])

        results.append({
            "Feature 1": col1,
            "Feature 2": col2,
            "Cramer's V": value,
            "k": k,
            "Strength": association_strength(value, k)
        })

association_results = pd.DataFrame(results)
display(association_results.sort_values("Cramer's V", ascending=False).round(3))
```

Berdasarkan analisis Cramer's V, pasangan fitur dengan asosiasi yang relatif tinggi antara lain:
- booking_channel - is_returning_guest memiliki nilai sebesar 0,460 dengan k=1 (Moderate).
- meal_package - booking_channel memiliki nilai sebesar 0,227 dengan k=3 (Moderate).
- room_category - booking_channel memiliki nilai sebesar 0,167 dengan k=4 (Moderate).

Ketiga pasangan tersebut menunjukkan bahwa booking_channel memiliki keterkaitan dengan status tamu, paket makanan, dan kategori kamar yang dipesan. Secara keseluruhan, tidak ditemukan asosiasi yang kuat antarfitur kategorikal.

```python
# Kolom yang seharusnya tidak bernilai negatif
non_negative_cols = [
    "num_adults",
    "num_children",
    "weekend_nights",
    "weekday_nights",
    "days_before_arrival",
    "prior_cancellations",
    "prior_completed_bookings",
    "nightly_rate_idr",
    "special_requests"
]

# Kolom yang hanya boleh bernilai 0 atau 1
binary_cols = [
    "needs_parking",
    "is_returning_guest",
]

def check_validity(df):
    checks = {}

    # Pemeriksaan nilai negatif
    for col in non_negative_cols:
        checks[f"{col}: Negative Values"] = (df[col] < 0).sum()

    # Pemeriksaan variabel binary
    for col in binary_cols:
        checks[f"{col}: Invalid Binary Values"] = (
            df[col].notna() & ~df[col].isin([0, 1])
        ).sum()

    if "is_canceled" in df.columns:
        checks["is_canceled: Invalid Binary Values"] = (
            df["is_canceled"].notna() & ~df["is_canceled"].isin([0, 1])
        ).sum()

    # Pemeriksaan total tamu
    total_guests = df["num_adults"] + df["num_children"]
    checks["Zero Total Guests"] = (total_guests == 0).sum()

    # Pemeriksaan validitas tanggal kedatangan
    date_cols = ["arrival_year", "arrival_month", "arrival_day"]
    date_parts = df[date_cols].rename(columns={
        "arrival_year": "year",
        "arrival_month": "month",
        "arrival_day": "day"
    })
    dates = pd.to_datetime(date_parts, errors="coerce")
    checks["Invalid Arrival Dates"] = (
        df[date_cols].notna().all(axis=1) & dates.isna()
    ).sum()

    return pd.Series(checks)

# Membandingkan train dan test
validity = pd.DataFrame({
    "Train": check_validity(train_df),
    "Test": check_validity(test_df)
})

display(validity)
```

### Outlier

```python
# Memilih fitur numerik, kecuali fitur kalender
excluded_cols = [
    "arrival_year",
    "arrival_month",
    "arrival_day"
]

numeric_cols = [
    col for col in train_df.columns
    if variable_types.get(col) == "integer"
    and col not in excluded_cols
]

def check_outliers(df):
    results = []

    for col in numeric_cols:
        data = df[col].dropna()

        Q1 = data.quantile(0.25)
        Q3 = data.quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        if IQR == 0:
            outlier_count = None
            outlier_pct = None
        else:
            outliers = (
                (data < lower_bound) |
                (data > upper_bound)
            )

            outlier_count = outliers.sum()
            outlier_pct = outliers.mean() * 100

        results.append({
            "Feature": col,
            "Min": data.min(),
            "Max": data.max(),
            "Q1": Q1,
            "Q3": Q3,
            "IQR": IQR,
            "Lower Bound": lower_bound,
            "Upper Bound": upper_bound,
            "Outlier Count": outlier_count,
            "Outlier (%)": outlier_pct,
            "Status": "IQR Not Applied" if IQR == 0 else "IQR Applied"
        })

    return pd.DataFrame(results)
```

# Preparation Pipeline <a name="6"></a>

---

<!-- rubric v3: fitted steps, fitted_prep_steps -->

The pipeline combines the preparation steps so they run in one consistent order.

```python
def build_preparation_pipeline():
    # Preprocessing numerik
    numerical_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ])

    # Preprocessing kategorikal
    categorical_pipeline = Pipeline([
        ("imputer", SimpleImputer(
            strategy="constant",
            fill_value="Unknown"
        )),
        ("encoder", OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        ))
    ])

    # Preprocessing biner
    binary_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent"))
    ])

    # Menggabungkan seluruh transformasi
    transformer = ColumnTransformer(
        transformers=[
            ("num", numerical_pipeline, numerical_cols),
            ("cat", categorical_pipeline, categorical_cols),
            ("bin", binary_pipeline, binary_cols)
        ],
        remainder="drop"
    )

    return Pipeline([
        ("preprocessing", transformer)
    ])
```

```python
preparation_pipeline = build_preparation_pipeline()

# Fit hanya pada training
X_train_pipeline = preparation_pipeline.fit_transform(X_train_raw)

# Transform validation dan test
X_val_pipeline = preparation_pipeline.transform(X_val_raw)
X_test_pipeline = preparation_pipeline.transform(X_test_raw)

# Memeriksa hasil
print("X_train:", X_train_pipeline.shape)
print("X_val:", X_val_pipeline.shape)
print("X_test:", X_test_pipeline.shape)
```

## Model Evaluation

<!-- figure palette: #6d2fa3 -->

This part compares the validation results of all models.

```python
models = {
   'knn': knn, 
   'dtl': dtl, 
   'svc': svc, 
   'lr': lr, 
   'gnb': gnb 
}
```

```python
def fit_predict(model, X_fit, y_fit, X_eval, method=None):
    X_resampled, y_resampled = resample(X_fit, y_fit, method)

    return deepcopy(model).fit(X_resampled, y_resampled).predict(X_eval)

def holdout(models, X_train, y_train, X_val, y_val, method=None):
    scores = {}

    for name, model in models.items():
        predicted = fit_predict(model, X_train, y_train, X_val, method)
        scores[name] = f1_score(y_val, predicted, average='macro')

    return pd.Series(scores, name='holdout').to_frame()
```

```python
holdout_table = holdout(models, X_train_final, y_train_final, X_val_final, y_val_final)
holdout_table
```

```python
def kfold_scores(model, X, y, method=None):
    X = np.asarray(X)
    y = np.asarray(y)
    folds = StratifiedKFold(n_splits=Settings.N_FOLDS, shuffle=True, random_state=Settings.SEED)
    scores = []

    for fit_index, eval_index in folds.split(X, y):
        predicted = fit_predict(model, X[fit_index], y[fit_index], X[eval_index], method)
        scores.append(f1_score(y[eval_index], predicted, average='macro'))

    return np.array(scores)

def kfold(models, X, y, method=None):
    rows = {}

    for name, model in models.items():
        scores = kfold_scores(model, X, y, method)
        rows[name] = {'kfold_mean': scores.mean(), 'kfold_std': scores.std()}

    return pd.DataFrame(rows).T
```

```python
kfold_table = kfold(models, X_train_final, y_train_final)
kfold_table
```

```python
def compare(holdout_table, kfold_table):
    return holdout_table.join(kfold_table).sort_values('kfold_mean', ascending=False)
```
